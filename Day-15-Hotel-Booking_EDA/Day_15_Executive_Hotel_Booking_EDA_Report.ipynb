## Data Cleaning and Preprocessing
Create a copy of the original dataset

```python
df_clean = df.copy()
```

```python
df_clean = df_clean.drop_duplicates()

print("Shape after removing duplicates:", df_clean.shape)
print("Duplicate rows remaining:", df_clean.duplicated().sum())
```

```python
missing_values = df_clean.isnull().sum()

missing_values[missing_values > 0].sort_values(ascending=False)
```

```python
df_clean['Hotel_Location'] = df_clean['Hotel_Location'].fillna('Unknown')
```

```python
df_clean['Children'] = df_clean['Children'].fillna(0)
```

```python
df_clean['Country'] = df_clean['Country'].fillna('Unknown')
```

```python
df_clean['Agent_ID'] = df_clean['Agent_ID'].fillna(0)
```

```python
df_clean['Company_ID'] = df_clean['Company_ID'].fillna(0)
```

```python
df_clean['Meal_Type'] = df_clean['Meal_Type'].fillna('Unknown')
```

```python
df_clean['ADR'] = df_clean['ADR'].fillna(df_clean['ADR'].median())
```

```python
df_clean['Satisfaction_Score'] = df_clean['Satisfaction_Score'].fillna(
    df_clean['Satisfaction_Score'].median()
)
```

```python
df_clean['Hotel_Type'] = (
    df_clean['Hotel_Type']
    .str.strip()
    .str.title()
)
```

```python
df_clean['Hotel_Type'].value_counts()
```

```python
df_clean['Market_Segment'] = (
    df_clean['Market_Segment']
    .str.strip()
    .str.title()
)
```

```python
df_clean['Market_Segment'].value_counts()
```

```python
df_clean['Meal_Type'] = (
    df_clean['Meal_Type']
    .str.strip()
    .str.upper()
)
```

```python
df_clean['Meal_Type'].value_counts()
```

```python
df_clean['Meal_Type'] = df_clean['Meal_Type'].replace({
    'B&B': 'BB'
})
```

```python
date_columns = [
    'Booking_Date',
    'Arrival_Date',
    'Reservation_Status_Date'
]

for col in date_columns:
    df_clean[col] = pd.to_datetime(df_clean[col])
```

```python
df_clean[date_columns].dtypes
```

```python
print("Missing Values:")
print(df_clean.isnull().sum())

print("\nDuplicate Rows:")
print(df_clean.duplicated().sum())

print("\nDataset Shape:")
print(df_clean.shape)
```

# Summary

The dataset was cleaned by removing duplicate records, handling missing values, standardizing inconsistent categorical values, and converting date-related columns to the appropriate datetime format. Numerical missing values such as ADR and Satisfaction Score were filled using the median, while categorical missing values were replaced with suitable categories such as “Unknown”. This preprocessing ensures that the dataset is suitable for reliable exploratory data analysis.

### Check for Outliers

```python
numerical_columns = [
    'Lead_Time_Days',
    'Weekend_Nights',
    'Weekday_Nights',
    'Total_Nights',
    'ADR',
    'Estimated_Revenue',
    'Satisfaction_Score'
]

df_clean[numerical_columns].describe()
```

```python
outlier_columns = [
    'Lead_Time_Days',
    'Total_Nights',
    'ADR',
    'Estimated_Revenue',
    'Satisfaction_Score'
]

plt.figure(figsize=(14, 8))

for i, col in enumerate(outlier_columns, 1):
    plt.subplot(2, 3, i)
    sns.boxplot(y=df_clean[col])
    plt.title(f'Box Plot of {col}')

plt.tight_layout()
plt.show()
```

```python
for col in outlier_columns:

    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df_clean[
        (df_clean[col] < lower_bound) |
        (df_clean[col] > upper_bound)
    ]

    print(f"{col}: {len(outliers)} potential outliers")
```

### Univariate Analysis

```python
plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=df_clean,
    x='ADR',
    y='Estimated_Revenue',
    alpha=0.6
)

plt.title('Relationship Between ADR and Estimated Revenue')
plt.xlabel('Average Daily Rate (ADR)')
plt.ylabel('Estimated Revenue')

plt.tight_layout()
plt.show()
```

### Cancellation Analysis

```python
cancellation_segment = (
    df_clean
    .groupby('Market_Segment')['Is_Canceled']
    .mean()
    .sort_values(ascending=False) * 100
)

plt.figure(figsize=(10, 6))

sns.barplot(
    x=cancellation_segment.index,
    y=cancellation_segment.values
)

plt.title('Cancellation Rate by Market Segment')
plt.xlabel('Market Segment')
plt.ylabel('Cancellation Rate (%)')

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

### Revenue Analysis

```python
segment_revenue = (
    df_clean
    .groupby('Market_Segment')['Estimated_Revenue']
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10, 6))

sns.barplot(
    x=segment_revenue.index,
    y=segment_revenue.values
)

plt.title('Total Estimated Revenue by Market Segment')
plt.xlabel('Market Segment')
plt.ylabel('Total Estimated Revenue')

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

### Monthly Booking Trend

```python
df_clean['Arrival_Month'] = df_clean['Arrival_Date'].dt.month_name()
```

```python
month_order = [
    'January', 'February', 'March',
    'April', 'May', 'June',
    'July', 'August', 'September',
    'October', 'November', 'December'
]
```

```python
monthly_bookings = (
    df_clean['Arrival_Month']
    .value_counts()
    .reindex(month_order)
)

plt.figure(figsize=(12, 6))

sns.lineplot(
    x=monthly_bookings.index,
    y=monthly_bookings.values,
    marker='o'
)

plt.title('Monthly Hotel Booking Trend')
plt.xlabel('Month')
plt.ylabel('Number of Bookings')

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```
