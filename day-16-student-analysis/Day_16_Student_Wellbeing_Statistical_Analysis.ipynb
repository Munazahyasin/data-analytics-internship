### Select the Variables for Statistical Analysis

```python
variables = [
    'Weekly_Study_Hours',
    'Average_Sleep_Hours',
    'Daily_Screen_Time_Hours',
    'Stress_Score',
    'Academic_Readiness_Score'
]
```

### Calculate Mean, Median and Mode

```python
for col in variables:
    print(f"\n--- {col} ---")
    print("Mean:", df[col].mean())
    print("Median:", df[col].median())
    print("Mode:", df[col].mode().iloc[0])
```

Calculate Range, Variance and Standard Deviation

```python
for col in variables:
    print(f"\n--- {col} ---")

    print("Range:", df[col].max() - df[col].min())
    print("Variance:", df[col].var())
    print("Standard Deviation:", df[col].std())
```

Calculate Q1, Q3 and IQR

```python
for col in variables:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    print(f"\n--- {col} ---")
    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
```

Which variable has the greatest variability?

```python
std_values = df[variables].std()

print(std_values)
print("\nGreatest variability:", std_values.idxmax())
```

Identify Outliers Using the IQR Method

```python
outlier_columns = [
    'Weekly_Study_Hours',
    'Daily_Screen_Time_Hours',
    'Commute_Time_Minutes',
    'Monthly_Discretionary_Spending'
]
```

```python
for col in outlier_columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ]

    print(f"\n{col}")
    print("Lower Bound:", lower_bound)
    print("Upper Bound:", upper_bound)
    print("Number of Outliers:", len(outliers))
```

Compare Mean and Median Before and After Removing Outliers

```python
print("Mean:", df['Daily_Screen_Time_Hours'].mean())
print("Median:", df['Daily_Screen_Time_Hours'].median())
```

```python
col = 'Daily_Screen_Time_Hours'

Q1 = df[col].quantile(0.25)
Q3 = df[col].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

df_no_outliers = df[
    (df[col] >= lower_bound) &
    (df[col] <= upper_bound)
]
```

```python
print("Mean after removing outliers:",
      df_no_outliers[col].mean())

print("Median after removing outliers:",
      df_no_outliers[col].median())
```

## Define Probability Events

```python
#Event A
A = df['Part_Time_Job'] == 'Yes'
```

```python
#Event B
B = df['Stress_Score'] >= 7
```

```python
#Event C
C = df['Scholarship'] == 'Yes'
```

```python
#Event D
D = df['Exercise_Days_Per_Week'] >= 3
```

Calculate Basic Probabilities

```python
P_A = A.mean()
P_B = B.mean()
P_C = C.mean()
P_D = D.mean()

print("P(A):", P_A)
print("P(B):", P_B)
print("P(C):", P_C)
print("P(D):", P_D)
```

Calculate P(A or B) and P(A and B)

```python
#A or B
P_A_or_B = (A | B).mean()

print("P(A or B):", P_A_or_B)
```

```python
#A and B
P_A_and_B = (A & B).mean()

print("P(A and B):", P_A_and_B)
```

Conditional Probability

```python
P_A_given_B = (A & B).sum() / B.sum()

print("P(A|B):", P_A_given_B)
```

```python
P_B_given_A = (A & B).sum() / A.sum()

print("P(B|A):", P_B_given_A)
```

Are Year 1 and Year 4 Mutually Exclusive?

Check Whether A and B Are Independen

```python
print("P(A and B):", P_A_and_B)

print("P(A) × P(B):", P_A * P_B)
```

Bayes' Theorem

```python
not_A = ~A

P_B_given_not_A = B[not_A].mean()

P_not_A = 1 - P_A

bayes_result = (
    P_B_given_A * P_A
) / (
    P_B_given_A * P_A +
    P_B_given_not_A * P_not_A
)

print("P(A|B) using Bayes Theorem:", bayes_result)
```

```python
print("Direct P(A|B):", P_A_given_B)
print("Bayes P(A|B):", bayes_result)
```

Normal Distribution Analysis

```python
mean_readiness = df['Academic_Readiness_Score'].mean()

std_readiness = df['Academic_Readiness_Score'].std()

print("Mean:", mean_readiness)
print("Standard Deviation:", std_readiness)
```

Calculate Z-Scores

```python
highest_score = df['Academic_Readiness_Score'].max()

highest_z = (
    (highest_score - mean_readiness)
    / std_readiness
)

print("Highest Score:", highest_score)
print("Z-score:", highest_z)
```

### Interpretation

The student with the highest academic readiness score is approximately 2.71 standard deviations above the mean.

Student with Lowest Score

```python
lowest_score = df['Academic_Readiness_Score'].min()

lowest_z = (
    (lowest_score - mean_readiness)
    / std_readiness
)

print("Lowest Score:", lowest_score)
print("Z-score:", lowest_z)
```

### Interpretation

According to the 68-95-99.7 empirical rule:

- Approximately 68% of students are expected to fall within 1 standard deviation of the mean.
- Approximately 95% are expected to fall within 2 standard deviations.
- Approximately 99.7% are expected to fall within 3 standard deviations.
