```python
import pandas as pd

df = pd.read_csv("Day8_Ecommerce_Sales_Dataset.csv")
df.head()
```

### Data selection

```python
#one column
df["Product"]
```

```python
#multiple columns
df[["Product", "Total_Sales"]]
```

Flitering

```python
#Show me only the rows that satisfy a condition
df[df["Total_Sales"] > 1000]
```

```python
#You can also combine conditions whit AND, OR
df[(df["Total_Sales"] > 1000) & (df["Category"] == "Electronics")]
```

Sorting

```python
#decending order
df.sort_values(by="Total_Sales", ascending=False)
#ascending order
df.sort_values(by="Total_Sales", ascending=True)
```

```python
_#top performing products
df.sort_values(by="Total_Sales", ascending=False).head(10)
```

## Groupby()

```python
df.groupby("Category")["Total_Sales"].sum()
```

```python
#Sales by city
df.groupby("City")["Total_Sales"].sum()
```

```python
#Sales by payment method
df.groupby("Payment_Method")["Total_Sales"].sum().sort_values(ascending=False)
```

### Multiple aggregation

```python
df.groupby("Category")["Total_Sales"].agg(["sum", "mean", "max", "min"])
```

Analyze quantity by category

```python
df.groupby("Category")["Quantity"].sum().sort_values(ascending=False)
```

Top performing catogories

```python
df.groupby("Category")["Total_Sales"].sum().sort_values(ascending=False).head()
```

Find Specific Order

```python
df[df["City"] == "Jammu"]
```

Combining filtering and sorting

```python
df[df["Category"] == "Electronics"].sort_values(
    by="Total_Sales",
    ascending=False
)
```
