## Find the common columns

---

### For example

Orders

CustomerID = 101

Customers
CustomerID = 101


Products
ProductID = 5

So we can connect the information.

```python
#Suppose your common column is CustomerID.
merged_data = pd.merge(
    orders,
    customers,
    on="Customer_ID",
    how="left"
)
```

```python
#Now connect the products.
#If the common column is Product_ID:
merged_data = pd.merge(
    merged_data,
    products,
    on="Product_ID",
    how="left"
)
```

## Demonstrate concat()

concat() is used to combine DataFrames along rows or columns. Here, two sample DataFrames are combined row-wise.

Important: concat() and merge() are different.

merge()

Used when datasets have a relationship through a common column.

```python
sample_orders = orders.head(5)
sample_orders_2 = orders.head(5)

combined_orders = pd.concat(
    [sample_orders, sample_orders_2],
    ignore_index=True
)

combined_orders
```

## Convert the Order Date

```python
merged_data["Order_Date"] = pd.to_datetime(
    merged_data["Order_Date"]
)
```

```python
#To check
merged_data["Order_Date"].dtype
```

```python
#Extract the month
merged_data["Month"] = merged_data["Order_Date"].dt.month
```

```python
#Extract the day
merged_data["Day"] = merged_data["Order_Date"].dt.day
```

```python
#Extract the day of the week
merged_data["DayOfWeek"] = merged_data["Order_Date"].dt.day_name()
```

```python
display(merged_data.head())
```

## Use apply()
One useful thing you can do is create a new column based on an existing column.

For example, if your dataset has a Price column:

```python
merged_data["Price_With_Tax"] = merged_data["Price"].apply(
    lambda x: x * 1.18
)

#This assumes an 18% tax calculation.
```

## Calculate a useful column

If your Products dataset contains a price and Orders contains quantity, you can create:

```python
merged_data["TotalSales"] = (
    merged_data["Quantity"] * merged_data["Price"]
)
```
