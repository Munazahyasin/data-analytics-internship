## Compare important variables using graphs

```python
sns.barplot(data=df, x="Region", y="Revenue")
plt.title("Average Revenue by Region")
plt.xticks(rotation=45)
plt.show()
```

```python
sns.barplot(data=df, x="Store_Type", y="Profit")
plt.title("Average Profit by Store Type")
plt.xticks(rotation=45)
plt.show()
```

```python
sns.barplot(data=df, x="Region", y="Customer_Rating")
plt.title("Average Customer Rating by Region")
plt.xticks(rotation=45)
plt.show()
```

## Correlation Analysis

```python
numerical_columns = df.select_dtypes(include=np.number)
```

```python
correlation_matrix = numerical_columns.corr()

correlation_matrix
```

```python
plt.figure(figsize=(10, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Matrix")
plt.show()
```
