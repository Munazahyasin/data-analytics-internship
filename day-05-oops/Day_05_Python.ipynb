# 🎓 Student Grade Management System

## Day 5 – Python Mini Project

This is a menu-driven Python application built using Object-Oriented Programming (OOP). The project allows users to add, view, search, and manage student records and grades.

```python
class Student:

    def __init__(self, name, roll_no, marks):
        self.name = name
        self.roll_no = roll_no
        self.marks = marks

    def get_grade(self):
        if self.marks >= 90:
            return "A"
        elif self.marks >= 75:
            return "B"
        elif self.marks >= 50:
            return "C"
        else:
            return "Fail"

    def display_details(self):
        print("\nStudent Details")
        print("Name:", self.name)
        print("Roll Number:", self.roll_no)
        print("Marks:", self.marks)
        print("Grade:", self.get_grade())
```

## Create the Management System class

```python
class GradeManagementSystem:

    def __init__(self):
        self.students = []

    def add_student(self):  # ✅ Inside the class
        try:
            name = input("Enter student name: ")
            roll_no = input("Enter roll number: ")
            marks = float(input("Enter marks: "))

            student = Student(name, roll_no, marks)
            self.students.append(student)

            print("Student added successfully!")

        except ValueError:
            print("Invalid input! Please enter valid marks.")
```

## Add a method to add students

```python
    def add_student(self):
        try:
            name = input("Enter student name: ")
            roll_no = input("Enter roll number: ")
            marks = float(input("Enter marks: "))

            student = Student(name, roll_no, marks)

            self.students.append(student)

            print("Student added successfully!")

        except ValueError:
            print("Invalid input! Please enter valid marks.")
```

##Add a method to view all students

```python
    def view_students(self):

        if not self.students:
            print("No student records found.")
            return

        for student in self.students:
            student.display_details()
```

## Add a search feature

```python
    def search_student(self):

        roll_no = input("Enter Roll Number to search: ")

        for student in self.students:
            if student.roll_no == roll_no:
                student.display_details()
                return

        print("Student not found.")
```

## Add a method to save data to a file

```python
    def save_to_file(self):

        with open("student_records.txt", "w") as file:

            for student in self.students:

                file.write(
                    f"Name: {student.name}, "
                    f"Roll No: {student.roll_no}, "
                    f"Marks: {student.marks}, "
                    f"Grade: {student.get_grade()}\n"
                )

        print("Student records saved successfully!")
```

## Complete menu-driven program

```python
system = GradeManagementSystem()

while True:

    print("\n===== STUDENT GRADE MANAGEMENT SYSTEM =====")
    print("1. Add Student")
    print("2. View All Students")
    print("3. Search Student")
    print("4. Save Records to File")
    print("5. Exit")

    choice = input("Enter your choice (1-5): ")

    if choice == "1":
        system.add_student()

    elif choice == "2":
        system.view_students()

    elif choice == "3":
        system.search_student()

    elif choice == "4":
        system.save_to_file()

    elif choice == "5":
        print("Thank you for using the system!")
        break

    else:
        print("Invalid choice! Please enter a number between 1 and 5.")
```
