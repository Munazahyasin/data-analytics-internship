# NumPy Practice Notebook

```python
import numpy as np

print("NumPy version:", np.__version__)
```

```python
#Creating 1D arrays
arr_1d = np.array([10, 20, 30, 40, 50])

print("1D Array:", arr_1d)

#Creating 2D arrays
arr_2d = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print("2D Array:")
print(arr_2d)

#Creating 2D arrays
arr_3d = np.array([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])

print("3D Array:")
print(arr_3d)
```

1.ndim → number of dimensions
2.shape → size of each dimension
3.dtype → data type of elements

```python
#Dimensions and data types
print("Dimensions:", arr_2d.ndim)
print("Shape:", arr_2d.shape)
print("Data type:", arr_2d.dtype)
```

```python
#Indexing
arr = np.array([10, 20, 30, 40, 50])

print(arr[0])
print(arr[2])
print(arr[-1])
```

```python
#Slicing
print(arr[1:4])
print(arr[:3])
print(arr[2:])
```

```python
#Reshaping
arr = np.arange(1, 13)

reshaped = arr.reshape(3, 4)

print(reshaped)
```

```python
#zeros()
zeros_array = np.zeros((3, 4))

print(zeros_array)

#ones()
ones_array = np.ones((2, 3))

print(ones_array)

#arange()
arr = np.arange(1, 11)

print(arr)

#linspace()
arr = np.linspace(0, 10, 5)

print(arr)
```

Mathematical operations

```python
a = np.array([10, 20, 30])
b = np.array([2, 4, 5])

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Square:", a ** 2)
```

Vectorized operations

```python
arr = np.array([1, 2, 3, 4, 5])

result = arr * 10

print(result)
```

Boolean masking

```python
arr = np.array([10, 25, 30, 15, 40, 5])

mask = arr > 20

print(mask)
print(arr[mask])
```

Broadcasting

```python
arr = np.array([10, 20, 30])

result = arr + 5

print(result)

#2D broadcasting
arr = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

result = arr + 10

print(result)
```

Aggregate functions

```python
arr = np.array([10, 20, 30, 40, 50])

print("Sum:", np.sum(arr))
print("Mean:", np.mean(arr))
print("Minimum:", np.min(arr))
print("Maximum:", np.max(arr))
print("Median:", np.median(arr))
```
